```python

"""
Global Interpreter Lock (GIL):  simplifies memory management and thread synchronization and avoids data corruption
GIL is a mechanism that ensures ONLY ONE THREAD EXECUTES PYTHON BYTECODE AT A TIME, even on MULTI-CORE PROCESSOR
This is done to prevent race conditions and data corruption in multi-threaded programs.
However, GIL can be a bottleneck for CPU-bound tasks that require a lot of computation.
To overcome this, Python provides multiprocessing module.

"""

import threading

def count():
    for i in range(10000000):
        pass

t1= threading.Thread(target=count)
t2= threading.Thread(target=count)

t1.start()
t2.start()

t1.join()
t2.join()

```

The provided code above creates two concurrent threads to run a CPU-bound loop, but due to Python's Global Interpreter Lock (GIL), they will execute sequentially on a single CPU core rather than in true parallel.

##### ⏱️ Code Execution Behavior
Sequential CPU Execution: 
1)      The count() function is entirely CPU-bound. Standard Python implementations (CPython) use a GIL. 
        This lock ensures only one thread executes Python bytecode at any given time.
     
The Result: 

2)      Instead of running both loops at the same time across two CPU cores, the operating system will constantly switch back and forth 
        between t1 and t2 on a single core. 
        This context-switching overhead actually makes the multi-threaded version slower than running a single loop of 20,000,000 sequentially.
  
 
##### 🛠️ How to Achieve True Parallelism

        If you want to utilize multiple CPU cores to speed up this computation, you should swap the threading module for the multiprocessing module
        This bypasses the GIL by spawning entirely separate Python processes.

###### Here is how you can rewrite your script using the multiprocessing documentation standards:

```python
import multiprocessing

def count():
    for i in range(10000000):
        pass

if __name__ == '__main__':
    # Creates two separate OS processes targeting separate CPU cores
    p1 = multiprocessing.Process(target=count)
    p2 = multiprocessing.Process(target=count)

    # Start execution side-by-side
    p1.start()
    p2.start()

    # Wait for both processes to complete
    p1.join()
    p2.join()
```
