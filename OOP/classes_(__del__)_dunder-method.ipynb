```python
class A:
    def __init__(self, x):
        self.x = x
        
    def __del__(self):
        print(f"Deleting {self.x}")

def create():                    # KEY CONCEPT: __del__() is called when an object is about to be destroyed(garbage collected)
    obj = A(99)
    return obj

obj = create()
del obj

import gc
gc.collect()



```
