##### ⏱️ Code Execution Behavior
Sequential CPU Execution: 
1)      The count() function is entirely CPU-bound. Standard Python implementations (CPython) use a GIL. 
        This lock ensures only one thread executes Python bytecode at any given time.
     
The Result: 

2)      Instead of running both loops at the same time across two CPU cores, the operating system will constantly switch back and forth

        This bypasses the GIL by spawning entirely separate Python processes.
