###  Polymorphism
##### Note: the situation in which the subclass is able to modify its superclass behavior (just like in the example) is called polymorphism. 
, which means that one and the same class can take various forms depending on the redefinitions done by any of its subclasses.

The method, redefined in any of the superclasses, thus changing the behavior of the superclass, is called virtual.

###### Each class's behavior may be modified at any time by any of its subclasses.

----------------------------------------------------------------------------
#####Let's analyze it:
    there are two classes, named 'One' and 'Two', while Two  is derived from One. So, 'Two' is a subclass of 'One'
    Nothing special. However, one thing looks remarkable - the do_it() method.

    the do_it()method is DEFINED twice: Once inside CLASS One and  then inside CLASS Two. 
    The essence of the example lies in the fact that it is INVOKED just once – inside CLASS 'One'.

The question is – which of the two methods will be INVOKED by the last two lines of the code?

#####The first invocation happens in CLASS 'One'
    seems to be simple, and it is simple, actually 
    – invoking doanything() from th OBJECT named 'one' will obviously activate the first of the methods.

#####The second invocation needs some attention. Keep in mind how Python finds class components. 
    The second invocation will launch do_it() from the CLASS Two, regardless of the fact that the invocation takes place 
    within the CLASS One.

##### Inheritance allows subclass Two to use the doanything() method from CLASS One
##### but since python reads classess from BOTTOM-TO-TOP it executes the do_it() method from subclass Two as it's the first found from bottom-to-top

----------------------------------------------------------------------------




```python
class One:
    def do_it(self):
        print("do_it from One")

    def doanything(self):
        self.do_it()


class Two(One):
    def do_it(self):
        print("do_it from Two")


one = One()
two = Two()

one.doanything()
two.doanything()
    
```

####Example One:

#####The loop runs .area() without needing to check if the object is a circle or a rectangle. 
#####Python figures out the correct method to run based on the object type
--------------------------------------------------------------------

```python

class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    
    def area(self):
        return self.width * self.length
    

# This loop or( can be a function too) treats both objects the same way:
shapes = [Circle(5), Rectangle(2, 3)]

for shape in shapes:
    print(f"Area of {type(shape).__name__} is {shape.area()} ")


# THis piece of code below is similar to teh one above in teh first explanation
#shapeC = Circle(5)
#shapeR = Rectangle(2,3)

#print(shapeC.area() )
#print(shapeR.area() )


#LearningMasterHub youtube
```
