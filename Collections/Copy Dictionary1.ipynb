####The use of update() : .update(other_data)
The update() method modifies a dictionary in-place by adding new key-value pairs or overwriting existing ones using data from another dictionary or iterable


- Returns: None (it mutates the original dictionary directly).
- Behavior: If a key exists, its value is overwritten. If a key is new, it is added


##### Way One to use update()
######Passing another dictionary

```python
dic1 = {"name": "Alice", "age": 25}
dic2 = {"age": 26, "city": "New York"}

dic1.update(dic2)
print(dic1)
# Output: {'name': 'Alice', 'age': 26, 'city': 'New York'}
```

##### Way Two to use update()
######Using Keyword Arguments

You can pass data directly as named arguments if your keys are valid Python identifiers (strings)

```python
dic3 = {"name": "Alice", "age": 25}

dic3.update(age=26, city="New York")
print(dic3)
# Output: {'name': 'Alice', 'age': 26, 'city': 'New York'}
```

##### Way Three to use update()
######Passing an Iterable of Key-Value Pairs
You can use a list of tuples, list of lists, or any iterable structured as pairs

```python
dic4 = {"name": "Alice", "age": 25}
tuple_data = [("age", 26), ("city", "New York")]

dic4.update(tuple_data)
# dic4.update( [("age", 26), ("city", "New York")] ) # It also works this ways. 


print(dic4)
# Output: {'name': 'Alice', 'age': 26, 'city': 'New York'}
```

--------------------------------------------------------------
#### Removing Dictionary Items
--------------------------------------------------------------

######1 del: removes an item using its key

```python
del dict1["Chicago"]
print(dict1)
```

######2 pop(): removes the item with the given key and returns its value

```python
dict1.pop("Milwaukee")
print(dict1)
```

######3  popitem(): removes and returns the last inserted key-value pair

```python
print(dict1.popitem())
print(dict1)
```

######4 clear(): removes all items from the dictionary

```python
dict1.clear()
print(dict1)
```
