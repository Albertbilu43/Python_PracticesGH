### Open/Close files
---------------------------------------------------------
With open() statement is the recommended standard for handling files because it automatically closes the file once the code block finishes executing. 

This prevents memory leaks and file corruption, even if an unexpected error occurs inside your code block.

##### Why Use 'with open()' Instead of 'open()' ?
Without the with block, you must remember to manually type file.close(). If your program crashes before reaching that line, the resource stays locked by system memory.

--------------------------------------------------------
##### Common File Modes

--Mode -----Purpose-----------Action

    "r"         Read (Default)   Opens a file for reading; throws an error if the file doesn't exist.
--------------------------------------------------------
    "w"         Write            Creates a new file or overwrites an existing file completely.
--------------------------------------------------------
    "a"         Append           Adds new data to the end of the file without deleting existing data.
--------------------------------------------------------
    "b"         Binary           Used for non-text files like images or PDFs (e.g., "rb" or "wb").

##### Code Examples

#### 1. Reading an Entire File
        Use the .read() method to load the entire contents of a file into a single string

```python
with open("example.txt", "r", encoding="utf-8") as file:
    data = file.read()
    print(data)
```

```python
with open("/Volumes/dev/spark_db/datasets/Python_PracticesData/data/test_orders.csv", "r", encoding="utf-8") as file:
    data = file.read()
    print(data)
```

#### 2. Reading a File Line-by-Line
        For large files, iterating through the file object using a for loop is highly efficient because it memory-buffers one line at a time.
        .

```python
with open("example.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())  # .strip() removes extra newlines
```

#### 3. Writing Data to a File
        Use "w" to create a file and write string data into it
        .

```python
with open("/Volumes/dev/spark_db/datasets/Python_PracticesData/data/test_orders.csv", "w", encoding="utf-8") as file:
    file.write("Hello, World!\n")
    file.write("This is a new line.")
```

----------------------------------------------------------------
#### 4. Appending Data to an Existing File
        Use "a" to add text without losing the original contents
        .

```python
# Example
with open("output.txt", "a", encoding="utf-8") as file:
    file.write("\nThis line is appended to the bottom.")
```

Looking at Cell 12, it already overwrote the file with "w" mode, so the CSV now contains only "Hello, World!\nThis is a new line." The user is trying to append to it in Cell 15.

The correct fix is to read the existing content, append the new line, and write it all back:

```python
# Note: UC Volumes do not support append mode - read, modify, and write back
with open("/Volumes/dev/spark_db/datasets/Python_PracticesData/data/test_orders.csv", "r", encoding="utf-8") as file:
    content = file.read()

with open("/Volumes/dev/spark_db/datasets/Python_PracticesData/data/test_orders.csv", "w", encoding="utf-8") as file:
    file.write(content + "\nThis line is appended to the bottom.")
```
