#### 1) None
None is a special constant used to represent the absence of a value or a null value. It is an object of its own data type (NoneType).

##### Common Uses of None
- Default Return Value: Functions that do not explicitly use a return statement automatically return None.
- Variable Initialization: Use it as a placeholder for variables that will receive data later.
- Default Arguments: Used in function definitions to signify optional parameters, especially to avoid issues with mutable default arguments (like lists).

How to Check for None
Always use the is or is not operators to check for None. Do not use ==.
- Correct: if value is None:
- Incorrect: if value == None:Code Examples


```python

# 1. Default return value
def greet(name):
    print(f"Hello, {name}")

result = greet("Alice")
print(result)  # Outputs: Hello, Alice

# 2. Optional argument placeholder
def append_to_list(element, target_list=None):
    if target_list is None:
        target_list = []
    target_list.append(element)
    return target_list
```

##### What the Full Code Does

A linked list is a chain of items where each item points to the next one (1 -> 2 -> 3). 

Reversing the list turns the pointers around so they point backward (3 -> 2 -> 1).

To do this inside the function, you usually change the arrows step-by-step using a loop (iterative) or small repeated steps (recursive) until the old last node becomes the new head.

```python


# Definition for singly-liked list
class ListNode:
    def __init__(self, val=0, next=None):
        self.val= val
        self.next = next

"""This incomplete code is the beginning of an iterative algorithm to reverse a singly linked list. The code sets up the initial pointers (prev and current) but is missing the logic inside the while loop body and the return statement needed to finish the job.Missing Code & Final ImplementationTo make this function work, you must save the next node, flip the pointer backward, and shift your tracker variables forward:

Step-by-Step Code Break Down
- prev = NoneInitializes a pointer to track the node before the current one. It starts as None because the original first node (the head) will become the new tail, meaning its .next must point to None.

- current = headInitializes a cursor pointer starting at the beginning of the list to handle traversal.

- while current:Starts a loop that will visit every node from left to right. The loop stops once current falls off the tail edge and becomes None.

- next_node = current.nextSaves a reference to the remaining unreversed list. If you do not do this step first, you lose access to the rest of the chain when you change the arrow in the next line.

- current.next = prev
  The actual reversal step. This points the current node's pointer backward to the node behind it instead of forward.
  
- prev = current & current = next_node
  Shifts both tracking pointers one position forward along the list so the loop can handle the next item.

- return prevWhen current becomes None, prev will be resting exactly on the very last node. Because the list is reversed, this last node is now the new head of the entire list.

"""

def reverseList(head: ListNode) -> ListNode:
    prev    = None
    current = head
    while current:
        nxt = current.next  # 1. Save the next node
        current.next = prev # 2. Reverse the link direction
        prev = current      # 3. Step prev forward
        current = nxt       # 4. Step current forward
    return prev             # 5. prev is now the new head

#Helper function to build Linked-list from list
"""This code creates a helper starting point for building a linked list, using dummy and current.dummy holds a fake first node to anchor the start of the list.current acts as a moving pointer to add new items.How It Worksdummy = ListNode(0): Makes a temporary node with a value of zero that sits at the very beginning.current = dummy"""

def buildLinkedList(values):
    dummy = ListNode(0)
    current = dummy
    for val in values:
        current.next = ListNode(val)
        current = current.next
    return dummy.next

#Helper function to convert Linked-list to list
"""This code starts a function to turn a linked list into a normal Python list, beginning by creating an empty container list named result.How It Worksdef Linked_List_to_List(head):: This creates a function that takes the first item (head) of a linked list.result = []: This makes a new, empty standard list to store the values as you go through the linked list.

while head:: This starts a loop that will keep going until it reaches the end of the linked list.head.next:: This moves the pointer to the next item in the linked list.head.val:: This adds the value of the current item to the result list.head = head.next:: This moves the pointer to the next item in the linked list.return result:: This returns the final result list"""

def Linked_List_to_List(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result

#Example
values = [1,2,3,4,5]
head = buildLinkedList(values)
reversed_head = reverseList(head)
print(Linked_List_to_List(reversed_head) )


```
