### Files & Folders

This Python code counts the total number of files and folders (directories) inside a specific folder on a computer, including everything hidden inside subfolders.
Here is a breakdown of how it works:

#####1. Setup and Path Definition
- import os: This loads Python's built-in operating system module. It allows Python to interact with your computer's file system.
- PATH = r'...': This specifies the target folder. The r before the string means "raw string," which tells Python to ignore backslashes (\) so they don't cause formatting errors.
- files = 0 / dirs = 0: These act as counters, starting at zero.

#####2. Scanning the Folder
- os.walk(PATH): This is the core function. It starts at the main folder and systematically "walks" through every single subfolder underneath it.
- for root, dirnames, filenames in ...: At every single stop, os.walk splits the contents into three parts:
    -   root: The current folder path.
    -   dirnames: A list of all folders inside the current folder.
    -   filenames: A list of all files inside the current folder.
    
#####3. Counting and Output
- files += len(filenames): This counts how many file names are in the current list and adds that number to the total.
- dirs += len(dirnames): This counts how many folder names are in the current list and adds that number to the total.print(...): Finally, it prints out the individual counts and adds them together for a grand total.


### RUN THIS CODE IN VS CODE as DATABRICKS MIGHT BRING SOME ERRORS

```python
import os

# Specify the path to count files and directories
PATH = r'C:\Users\ale_r\Downloads'

files = 0
dirs  = 0

for root, dirnames, filenames in os.walk(PATH):
    files += len(filenames)
    dirs += len(dirnames)

print(f'Number of files: {files}')
print(f'Number of directories: {dirs}')
print(f'Totals: {files + dirs }')      

```
