```python
class Level1:
    variable_1 = 100
    def __init__(self, nivel1):
        self.nivel1= nivel1
        self.var_1 = 101

    def fun_1(self):
        return 102


class Level2(Level1):
    variable_2 = 200
    def __init__(self, nivel1, nivel2):
        super().__init__(nivel1)
        self.nivel2= nivel2
        self.var_2 = 201
        
    
    def fun_2(self):
        return 202


class Level3(Level2):
    variable_3 = 300
    def __init__(self, nivel1, nivel2, nivel3):
        super().__init__( nivel1, nivel2)
        self.nivel3= nivel3        
        self.var_3 = 301

    def fun_3(self):
        return 302


obj = Level3(1,2,3)

print(obj.variable_1, obj.var_1, obj.fun_1())
print(obj.variable_2, obj.var_2, obj.fun_2())
print(obj.variable_3, obj.var_3, obj.fun_3())
    
```
